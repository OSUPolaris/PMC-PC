```python
#Let's try to model the polymer under force, using a similar structure to the
# random walk in 3-d we did last time

import random
import numpy as np #We'll use the numpy module as it allows us to do some cool stuff
import matplotlib.pyplot as plt
```

```python
#Let's explore another ways to generate random numbers, because they might
# be easier for random walks, especially with chances that aren't 50-50

chance_left = 0.4
chance_right = 0.6

choice_array = ['left', 'right']
weight_array = [chance_left, chance_right]
choice = random.choices(choice_array, weights = weight_array)
print(choice)
```

```python
#Let's confirm we are getting statistics we like
len_test = 10000
num_r = 0
for i in range(10000):
  choice = random.choices(choice_array, weights = weight_array)
  if choice == ['right']: #Note that the output is still an array
    num_r += 1
num_l = len_test - num_r
plt.bar(['right', 'left'], [num_r, num_l])
print(f"num right: {num_r}, num left: {num_l}")
```
