# Polymer in a confined space
One more thing that is interesting is how polymers configure themselves inside of cells. For example, DNA is generally confined to the nucleus. If we were to model this as a random walk, you could imagine that when it gets close to the barrier, certain choices are no longer available. This will ensure that the random walk stays within a certain region. In our case, we're going to stick to a spherical region. We will also stick to a certain size of the region, and vary the length of the random walk.
## Questions
1. What do you think $<R>$ will be for a random walk like this? Would it change if you choose a non-symmetrical region?
2. What do you predict the graph for $<R^2>$ vs N to look like? Think about how the graph would look when $N*a$ (The number of steps times the step size) is much smaller or much longer than the radius of the sphere.

```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
#Here is a function that will return true if a location is in our region, and
# false if it is outside of the region
def spherical_region(_x, _y, _z, _rad):
  if (_x**2 + _y**2 + _z**2) <= _rad**2:
    return True
  return False
```

```python
Ns = np.linspace(10, 200, 100)
R_squared_agg = np.zeros(len(Ns))
#One thing we haven't done much is really ramp up how many sets of walks we do.
# You'll also want to go back later and apply this to your other plots if they
# are going to go on your poster, but here we can really make the plot look
# continuous.

# For every different N:
  # Do 10,000 random walks
    # Do N steps every walk
      #To implement barrier, when you make a step outside of it, go back instead

N_walks = 1000
a = 1
rad = 10

for i,j in enumerate(Ns):
  for k in range(N_walks):
    x = 0
    y = 0
    z = 0
    l = 0 #Here l is where we are in the walk. There is a reason why, because it
          # Will be easier to go backwards when we end up outside the barrier.
    while (l < j):
      choice_array = ['left', 'right', 'up', 'down', 'forward', 'back']
      weight_array = [1./6., 1./6., 1./6., 1./6., 1./6., 1./6.]
      choice = random.choices(choice_array, weights = weight_array)
      if choice == ['left']:
        x -= a
      elif choice == ['right']:
        x += a
      elif choice == ['up']:
        y += a
      elif choice == ['down']:
        y -= a
      elif choice == ['forward']:
        z += a
      elif choice == ['back']:
        z -= a

      #Now check if inside barrier, and only tick l up if we are inside,
      # otherwise, go back.

      if spherical_region(x, y, z, rad):
        l += 1
      else:
        if choice == ['left']:
          x += a
        elif choice == ['right']:
          x -= a
        elif choice == ['up']:
          y -= a
        elif choice == ['down']:
          y += a
        elif choice == ['forward']:
          z -= a
        elif choice == ['back']:
          z += a

    R_squared_agg[i] += x**2+y**2+z**2
  R_squared_agg[i] /= N_walks
  R_squared_agg[i] = np.sqrt(R_squared_agg[i])
```

```python
plt.plot(Ns, R_squared_agg)
plt.xlabel('N')
plt.ylabel('sqrt(<R^2>)')
```

Let's also plot what we know is the answer for an unconstrained random walk. Remember that our answer was $\sqrt{<R^2>}=\sqrt{N}a$.

```python
def unconstrained(_N, _a):
  return np.sqrt(_N)*_a

plt.plot(Ns, unconstrained(Ns, a))
plt.plot(Ns, R_squared_agg)


plt.xlabel('N')
plt.ylabel('sqrt(<R^2>)')
```

One more thing to think about is this: When the random walk is much larger than the barrier, it is almost like the random walk goes to a uniformly random spot within the sphere. It turns out if you do the math, if you pick a random spot within the sphere, the average distance from the origin will be 0.75 times the radius. Let's also plot this.

```python
plt.plot(Ns, unconstrained(Ns, a))
plt.plot(Ns, (Ns*0) + 0.75*rad, linestyle = '--') #here we just multiply the y axis by 0 so we get an array of the appropriate length
plt.plot(Ns, R_squared_agg)


plt.xlabel('N')
plt.ylabel('sqrt(<R^2>)')
```
