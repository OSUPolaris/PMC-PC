### Lists

Lists are the most forgiving container in Python. Basically, we can shove whatever we want into a list, be they different data types, or even lists within lists. Of course, the usefulness of non-uniform containers becomes limited - the advantage to storing lots of numbers in a list is that you can then perform operations on them all without worrying that some won't work. 

Here's how we define a list:

```python
list_1 = ['a string', 5, 6.0, True, [5,6,6]]
```

The list above is kind of a mess - you'd almost never want a list to contain such a variety of things in it, but I wanted to highlight that, in principle, Python doesn't care what you stick inside a list.
```python
count_to_10 = range(1,11)
print(count_to_10)
```

What I've done above is run the "range()" function, which generates a list of numbers counting up. The form of the function is range(start, stop, step), where "start" is inclusive and "stop" is exclusive (e.g. [1,11) ). If I wanted 0-9, I would do

```python
range(10)
```

using the shortcut that if only 1 argument is used, "start" is assumed to be 0 and "step" is assumed to be 1 (it has to be an integer).

Okay, so the other thing students are always interested in is the standard deviation from the mean - this basically will tell them whether they get an A, B, C, D, or F on the test assuming you curve. Let's be nice, and assume we are going to curve the average to a flat B (85%). The formula for a standard deviation is 
$$ 
s = \sqrt{\frac{\sum_{1}^{N}(x_i - \mu)^2}{N-1}}
$$

where $\mu$ is the average and N is the total number of scores.

We already know how to get N, and we know what $\mu$ is as well. So to calculate this, we need to know how to calculate the quantity on the top of the fraction. This is actually kind of tricky with the methods we have on hand, so I'm going to introduce a new concept: Numpy (numerical python) arrays. I'm going to get into these in detail in part 2 of the bootcamp, but for now, see the example below for elucidation on why we're about to use them:

```python
import numpy as np
arr_version = np.array(scores)
print(scores-1)
```

Okay, so I can't subtract an integer from a list. What if I try the array version?

```python
print(arr_version-1)
```

If you look, you should see that each of those scores is the original score with one subtracted off it. Your spidey senses should be tingling then for how we can leverage this functionality to calculate our STD. In the cell below, fill in the variable I'm calling "top_frac" to calculate this quantity:
$$
\sum_{i=1}^N (x_i - \mu)^2
$$

Notice here that you don't have to actually calculate it one by one - if we first compute a single array that represents each score with the mean subtracted off and then that value squared, then we finish off top_frac just by summing up that array as we've done before. Feel free to use my variable "arr_version".

```python
top_frac = sum((arr_version - average_score)**2) #your code here
print(top_frac)
```

With that done, we can easily apply the formula to get the final STD - **Hint:** the function np.sqrt() will be useful here.

```python
STD_scores = np.sqrt(top_frac/ (len(scores) - 1))#your code here
print(STD_scores)
```

## Curve Fitting Examples
It's always a good idea to perform a "sanity check" on your code - is your chosen algorithm doing what it should be doing? One way to do this is to first simulate the motion of particle whose equation of motion has an exact, analytic solution. Fitting the shape of this analytic solution to the "data" produced by your simulation is a useful way to check the stability and accuracy of your simulation (this is called "parameter retrieval"). Once you've confirmed your code works for the simple case, it's likely to give you an accurate result for more complex situations. 

Therefore, knowing how to fit some basic curves to data will be useful this semester! This notebook shows some basic examples of fitting curves to randomly generated data. The SciPy tool "curve_fit" will be very useful for our purposes.

```python
from scipy.optimize import curve_fit
```

```python
npts = 20 # Create twenty fake data points (note, this is easy to change!)

# First, let's do a simple straight-line fit: y = mx + b
# Define m and b first, so we can see how well curve_fit is able to recover our inputs
m = 1.5
b = 3.7

x_data = np.linspace(0,10,npts)
y_data = m * x_data + b

plt.figure(figsize=(6,6))  # Customize the size of your figure
plt.plot(x_data,y_data,'ok')
plt.xlabel('x',fontsize=16)  # Also an example of how to change the font size of your plot labels
plt.ylabel('y',fontsize=16)
plt.title('No Noise',fontsize=18);
```

```python
# The data above are a perfect straight line - fitting a curve to this isn't interesting
# Let's make it more realistic: add some "noise" to the data

y_data = m * x_data + b + np.random.normal(0,1.5,npts)

plt.figure(figsize=(6,6))  # Customize the size of your figure
plt.plot(x_data,y_data,'ok')
plt.xlabel('x',fontsize=16)  # Also an example of how to change the font size of your plot labels
plt.ylabel('y',fontsize=16)
plt.title('With Noise',fontsize=18);
```

```python
# We're ready to fit a straight line to our noisy data!

# First, define the function to fit a line:
def linefit(x,m,b):
    return m*x+b

params , param_cov = curve_fit(linefit,x_data,y_data)

# Create a new, pretty plot showing the fit results
plt.figure(figsize=(6,6))  # Customize the size of your figure
plt.plot(x_data,y_data,'ok',label='data')
plt.plot(x_data,params[0]*x_data+params[1],'-r',linewidth=2,label='fit')
plt.plot(x_data,m*x_data+b,'--b',label='true')
plt.legend(fontsize=12)
plt.xlabel('x',fontsize=16)  # Also an example of how to change the font size of your plot labels
plt.ylabel('y',fontsize=16)
plt.title('Linear Fit (With Noise)',fontsize=18);

# Print out the best-fit parameters, with pretty formatting
print('Best-fit slope = {:.2f} (input value = {:.2f})'.format(params[0],m))
print('Best-fit intercept = {:.2f} (input value = {:.2f})'.format(params[1],b))
```

```python
# What is the uncertainty on our fit parameters?
# Calculate with the covariance matrix (saves as param_cov)
params_err = np.sqrt(np.diag(param_cov))

# Print out the best-fit parameters, again, with uncertainties and pretty formatting
print('Best-fit slope = {:.2f} +/- {:.2f} (input value = {:.2f})'.format(params[0],params_err[0],m))
print('Best-fit intercept = {:.2f} +/- {:.2f} (input value = {:.2f})'.format(params[1],params_err[1],b))
```

### A slightly more complicated example: fitting a sinusoid

```python
npts = 20 # Create twenty fake data points (note, this is easy to change!)

# Let's create and fit some fake data that follow a sinusoidal function, with noise
# Note that I'm purposefully naming my independent variable "time"
A = 2
omega = 0.5 * np.pi
phase = 0.25
offset = 1

time = np.linspace(0,10,npts)
y_data = A * np.sin(omega * time + phase) + offset + np.random.normal(0,0.35,npts)

# Define the function to fit to the data
def FitSine(t,a,b,c,d):
    return a * np.sin(b*t + c) + d

params, param_cov = curve_fit(FitSine,time,y_data)#optimize.curve_fit(FitSine,time,y_data)
params_err = np.sqrt(np.diag(param_cov))

plt.figure(figsize=(6,6))  # Customize the size of your figure
plt.plot(time,y_data,'ok',label='data')
plt.plot(time,params[0]*np.sin(params[1]*time + params[2]) + params[3],'-r',label='fit')
plt.plot(time,A*np.sin(omega*time+phase)+offset,'--b',label='true')
plt.legend()
plt.xlabel('x',fontsize=16)  # Also an example of how to change the font size of your plot labels
plt.ylabel('y',fontsize=16)
plt.title('Fake Data with Fit',fontsize=18);

print('Best-fit amplitude = {:.2f} +/- {:.2f} (input value = {:.2f})'.format(params[0],params_err[0],A))
print('Best-fit omega = {:.2f} +/- {:.2f} (input value = {:.2f})'.format(params[1],params_err[1],omega))
print('Best-fit phase = {:.2f} +/- {:.2f} (input value = {:.2f})'.format(params[2],params_err[2],phase))
print('Best-fit offset = {:.2f} +/- {:.2f} (input value = {:.2f})'.format(params[3],params_err[3],offset))
```

Wait, what happened above? Clearly the fit did not recover the correct answer! This can sometimes happen when your data are sparse. Generally, there are two ways around this: increase the number of data points (or "steps") in your independent variable, or start the fit with some initial "guesses" for what the best-fit parameters will be. The former is easy to execute when you deal with simulated data (like in this class), while the latter is often the only option when you are dealing with observational data. Let's see what happens when we do it both ways.

```python
npts = 100 # Create twenty fake data points (note, this is easy to change!)

# Let's create and fit some fake data that follow a sinusoidal function, with noise
# Note that I'm purposefully naming my independent variable "time"
A = 2
omega = 0.5 * np.pi
phase = 0.25
offset = 1

time = np.linspace(0,10,npts)
y_data = A * np.sin(omega * time + phase) + offset + np.random.normal(0,0.35,npts)

# Define the function to fit to the data
def FitSine(t,a,b,c,d):
    return a * np.sin(b*t + c) + d

params, param_cov = curve_fit(FitSine,time,y_data)
params_err = np.sqrt(np.diag(param_cov))

plt.figure(figsize=(6,6))  # Customize the size of your figure
plt.plot(time,y_data,'ok',label='data')
plt.plot(time,params[0]*np.sin(params[1]*time + params[2]) + params[3],'-r',label='fit')
plt.plot(time,A*np.sin(omega*time+phase)+offset,'--b',label='true')
plt.xlabel('x',fontsize=16)  # Also an example of how to change the font size of your plot labels
plt.ylabel('y',fontsize=16)
plt.title('Fake Data with Fit: Increased # of Points to Fit',fontsize=18);

print('Best-fit amplitude = {:.2f} +/- {:.2f} (input value = {:.2f})'.format(params[0],params_err[0],A))
print('Best-fit omega = {:.2f} +/- {:.2f} (input value = {:.2f})'.format(params[1],params_err[1],omega))
print('Best-fit phase = {:.2f} +/- {:.2f} (input value = {:.2f})'.format(params[2],params_err[2],phase))
print('Best-fit offset = {:.2f} +/- {:.2f} (input value = {:.2f})'.format(params[3],params_err[3],offset))
```
