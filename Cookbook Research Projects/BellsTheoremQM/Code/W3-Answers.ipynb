```python
import numpy as np
import scipy.linalg as la
```

At the end of the linear algebra notebook, I mentioned that quantum mechanics is all about solving the Schrödinger equation:

$$ \left(-\frac{\hbar^2}{2m} \nabla^2 + V(\vec{r})\right)\Psi(\vec{r},t) = -i\hbar \frac{\partial}{\partial t} \Psi(\vec{r},t) $$

which in this form seems a very daunting task: a second-order, linear differential equation which generally depends on all three dimensions of space and time. Luckily, for the sake of our project, we don't need to worry about solving the equation in this form. Instead, we will work with a much more palatable form of the Schrödinger equation:

$$ \newcommand{\ket}[1]{\left|{#1}\right\rangle} \newcommand{\bra}[1]{\left\langle{#1}\right|}
\newcommand{\braket}[2]{\left\langle{#1}\middle|{#2}\right\rangle}
\mathcal{H} \ket{\Psi} = E \ket{\Psi} $$

These two equations aren't exactly equivalent; the second version no longer depends on time, and we'll only work on problems where the physics is constant in time to avoid that complication. Each part of this new equation is interpreted as follows:

1. $\mathcal{H}$ is called the **Hamiltonian** and can be thought of as a matrix/operator which is equivalent to all the things in parentheses on the left side of the first equation. This operator is used to determine the energy of a quantum state. We'll see exactly how this works shortly.
2. $\ket{\Psi}$ is called the **wave function** and can be considered as a vector which is equivalent to the function $\Psi(\vec{r},t)$ in the first equation. The wave function encodes all the information for a quantum system, and we use operators to extract information; for example, if I have an electron in a hydrogen atom, the wave function can be used to determine the position, momentum, spin, and other measurements for the electron.
3. $E$ is (as you may have guessed) the total energy of the system. 

The form of this Schrödinger equation is the same as the form of the eigenvalue equations from before: a matrix, multiplied by a vector, equals a number times the same vector. In this way, the energy is an eigenvalue of the Hamiltonian operator. When we talk about "solving" the Schrödinger equation, what we want to do is determine the eigenvalues and eigenvectors of the Hamiltonian operator. 

## Quantum State Vectors

The overall state of a quantum system is described by its wave-function, often written as $\Psi(\vec{r},t)$ or shorthand as $\ket{\Psi}$, and can be treated as a column vector where each component corresponds to a different quantum state. I can also have the state $\bra{\Psi}$, which is $\ket{\Psi}$ written as a row vector (technically, it's the Hermitian conjugate). The column vector $\ket{\Psi}$ is called a "ket" while the row vector is called a "bra"; putting them together, as in $\braket{\Psi}{\Psi}$, forms a "bra-ket" or bracket. Remember from last time, multiplying a row vector by a column vector is equivalent to taking the inner (dot) product of the two vectors. You can also have a "ket-bra" $\ket{\Psi}\bra{\Psi}$ which is the outer product of the vectors, but we likely won't need to use this.

As an example, consider a quantum system which consists simply of an arrow which points up or down. I can assign each state a vector in a 2-dimensional space (1 dimension for each state). I'll define the "up" state to be the vector 
$$ \ket{\uparrow} = \pmatrix{1 \\ 0} $$

and the "down" state will be 

$$ \ket{\downarrow} = \pmatrix{0 \\ 1}. $$

These two vectors are my **basis vectors**, meaning any other state can be written as some combination of these two. You can also check that they are orthogonal (perpendicular) by calculating their inner product:

$$ \braket{\uparrow}{\downarrow} = \pmatrix{1 & 0}\pmatrix{0 \\ 1} = 0. $$

You can visualize this as a vector in the 2D Cartesian plane, where "up" is the x-component and "down" is the y-component. In this way, I can also make a state which points anywhere else in the plane; for example, 

$$ \ket{\phi} = \frac1{\sqrt{2}} \pmatrix{-1 \\ 1}. $$

The factor of $\frac1{\sqrt{2}}$ comes from an important property of quantum states called **normalization**; since each component of the vector corresponds to a particular basis vector,
this is equivalent to saying that the "length" of our state vector must be 1. Mathematically, this means that the square of the inner product of the state with itself is 1:

$$ \left|\braket{\phi}{\phi}\right|^2 = \left(\frac1{2} + \frac1{2}\right)^2 = 1. $$

We'll see in the next section how this guarantees that the total probability of measuring each outcome adds up to 1 (which is necessary for probabilities to make sense).

This quantum state is in a **superposition** of both the "up" and "down" states; in the language of the popular Schrödinger's cat, it is both "alive" and "dead". It can't be both at once though, can it?

### Quantum Measurements

Whenever we measure something in the lab, we always get some definite answer; for example, if we measure our quantum arrow, we should always get either up or down, not both, as an answer. So how do we know which one we'll measure when we have the state $\ket{\phi}$ above? The components of our state vector have the answer. 

Supposed I have an arbitrary quantum state $\ket{\Psi}$. I want to know the probability that when I measure my state, I will measure a particular outcome $\ket{\alpha}$. 

**Definition:** the probability that I will measure the outcome $\ket{\alpha}$ from a measurement of the state $\ket{\Psi}$ is the square of the inner product of the two states:

$$ P(\alpha) = \left|\braket{\alpha}{\Psi}\right|^2 $$

The absolute value signs are crucial, as some quantum states can in general have negative or complex values; taking the square of the magnitude makes sure that no matter what, the probability is a nonnegative number and therefore makes sense. 

For our example of the arrow system: The probability for you to measure "up" when you measure the state $\ket{\phi}$ is given as:

$$ P(\uparrow) = \left|\braket{\uparrow}{\phi}\right|^2 = \left(-\frac1{\sqrt{2}}+0\right)^2 = \frac1{2}, $$

and the probability for "down" is 

$$ P(\downarrow) = \left|\braket{\downarrow}{\phi}\right|^2 = \left(0 + \frac1{\sqrt{2}}\right)^2 = \frac1{2}. $$

It's important to note that if I add the probabilities of "up" and "down" that I get 1; this is because my state is normalized correctly. Each component of the state vector corresponds to the probability of measuring that basis vector; adding up these probabilities is equivalent to adding up the components of the vector, squared, to determine the length. 

If we do our measurement of the arrow system 100 times, we expect that we'll observe "up" about 50 times and "down" the other 50 times, give or take for some statistical fluctuations. It's certainly possible that for 10 measurements, we might see 7 up and 3 down, but as we repeat the measurement more and more times, it should even out closer and closer to being exactly 50/50.

This is one of the crucial ideas in all of quantum mechanics: classical physics is **deterministic**, meaning that the same experiments with exactly the same setups will always yield the same results, while quantum physics is **probabilistic**. Repeating the same experiment may yield different results, but those results follow a definite probability distribution. Einstein hated and disagreed with this idea, prompting his famous comment, "God does not play dice," but unfortunately, every quantum experiment humanity has ever done confirms that this is how the world seems to work. 

#### Practice Problem: 

Consider a new quantum system: the stoplight system. Our system can be in three different states: red $\ket{\mathsf{R}}$, yellow $\ket{\mathsf{Y}}$, and green $\ket{\mathsf{G}}$. We'll define the basis vectors as follows:

$$ \ket{\mathsf{R}} = \pmatrix{1 \\ 0 \\ 0};\ \ket{\mathsf{Y}} = \pmatrix{0 \\ 1 \\ 0};\ \ket{\mathsf{G}} = \pmatrix{0 \\ 0 \\ 1}. $$

Suppose you have prepared a quantum state of the form 

$$ \ket{\phi} = A \pmatrix{3 \\ 4 \\ 5} $$ 

1. Determine the value of the constant $A$ so that the state $\phi$ is normalized properly.
2. For each color, determine the probability of measuring that color. 

Feel free to do this calculation by hand, or with Python (or both, if you're really feeling it).

```python
phi0 = np.array([3,4,5]);
A = np.sqrt(1/sum(phi0**2));
print(A)

phi = A*phi0;
print(sum(phi**2))
```

```python
P_R = phi[0]**2;
P_Y = phi[1]**2;
P_G = phi[2]**2;

print("Probability of measuring Red: ", P_R);
print("Probability of measuring Yellow: ", P_Y);
print("Probability of measuring Green: ", P_G);
```

### Infinite-Dimensional Space

When we write a quantum state as a vector, each component corresponds to a possible measurement of the system. This is easy enough to parse for systems which have a discrete, countable number of measurements; however, in your quantum mechanics courses, you'll deal with wave functions that tell you the probability of measuring continuous variables. For example, position is a continuous variable with an uncountably infinite number of different values, so we'd have to write our vector with an uncountably infinite number of components (this is impossible). Instead, we rely on writing the state as a wave function $\Psi(\vec{r})$; instead of having different components for the probability of each state, the value of the function  will be larger at positions which are more probable. The probability of measuring the particle in a particular region of space is found with an integral of $|\Psi|^2$ over that region. To normalize the wave function, instead of making sure all the probabilities add up to 1, we make sure that the integral of $|\Psi|^2$ over all of space is 1.

Almost all of the physics in our project will rely on 2-state systems, and 2-component vectors, so you won't need to do any difficult integrals here, but I just wanted to make you aware of this so that you'll hopefully recognize it later on in a quantum class.


## Quantum Operators

As I mentioned before, quantum operators like the Hamiltonian in the Schrödinger equation can be thought of as a square matrix which multiplies a quantum state vector. While the state vector encodes information about one particular state of a system, an operator encodes all of the possible states the system could be in and all the possible values I can measure. All this information is stored in the eigenvectors and eigenvalues, which we can easily find using Python, so once we write down an operator for some physical system, we're just a short calculation away from uncovering a lot of the details of what we can see in an experiment. 

To see how all this works, lets return to the stoplight example. We have three basis states, called Red $\ket{\mathsf{R}}$, Yellow $\ket{\mathsf{Y}}$, and Green $\ket{\mathsf{G}}$. These basis states are the eigenvectors of some operator I'll call $\mathcal{H}$; this will be the Hamiltonian of my system which tells me the energy of each state. Each eigenvector has a corresponding eigenvalue which is the energy I'll measure in that state; let's call them $E_R, E_Y, E_G$. Then my Hamiltonian will be something like 

$$ \mathcal{H} = \pmatrix{E_R & 0 & 0 \\ 0 & E_Y & 0 \\ 0 & 0 & E_G} $$

Suppose $E_R=-2, E_Y=1, E_G=4$ in whatever units of energy you like (Joules, eV, calories, who cares). Because $E_R$ is the lowest value, we call $\ket{\mathsf{R}}$ the **ground state**; $\ket{\mathsf{Y}}$ is the first **excited state** and $\ket{\mathsf{G}}$ is the second.

### Practice Problem

Create an matrix (array) in Python for $\mathcal{H}$ using the values above for the energies. Referring back to the linear algebra notes, calculate the eigenvalues and eigenvectors of $\mathcal{H}$ to verify that our basis states are indeed the eigenvectors of $\mathcal{H}$ with the appropriate corresponding eigenvalues.

```python
H = np.array([ [-2,0,0],[0,1,0],[0,0,4] ]);

eigvals, eigvecs = la.eig(H);
print(eigvecs)
print(eigvals)
```

### Measurements vs. Expectation Values

Now we know that we have three basis states, each with a different energy value. How do I know the energy of the state $\ket{\phi}$ from before? This question isn't very well defined in quantum mechanics, because my state is made up of components of all three basis states, with each component corresponding to a different energy. There are two ways we can rephrase this question that give a sensible answer.
 1. What is the probability that a measurement on my state will yield a particular energy value?
 2. What is the average value of the energy that I will get if I measure this state repeatedly?

**Question 1:** This is actually the same as the second part from the stoplight practice problem. In that problem, you calculated the probability that you would measure a given state, say $\ket{\mathsf{R}}$, by taking the inner product of $\ket{\phi}$ with that state, then squaring the result. This gives the probability that your measurement of the energy of $\ket{\phi}$ would be $E_R$. This means that if I repeated my experiment 100 times, I wouldn't get the same energy every time, but I would get one of $E_R$, $E_Y$, or $E_G$ each time, and the amount of times I measured each one would line up with the probabilities. This is part of the weirdness of superposition: the same quantum state can measure different values in an experiment, because the state is made up of components from different eigenvectors. When I measure one particular value, you'll sometimes hear people say that the wave function "collapses" to that corresponding basis state.

**Question 2:** When we naively ask "What's the energy?", Question 2 is often what we really mean to ask. Experimentally, the **expectation value** of an operator, such as energy, is simply the average value: if I repeat an experiment 100 times, I record the measured value each time, then add them all up and divide by 100. The expectation value of an operator $\mathcal{O}$ is often written as $\langle \mathcal{O} \rangle$, which is shorthand for the calculation:

$$ \langle \mathcal{O} \rangle = \bra{\phi} \mathcal{O} \ket{\phi}. $$

Reading from right to left, this means I start with my state $\ket{\phi}$ and multiply it by the operator $\mathcal{O}$ to get a new state. I then take the inner product of the new state with $\bra{\phi}$ to get a scalar (number) answer. This works because the components of $\ket{\phi}$ correspond to the probability of measuring a given eigenvalue, and including $\mathcal{O}$ in the center multiplies the probabilities by their corresponding measurement values, making this calculation essentially just a weighted average. 

#### Practice Problem

Find the expectation value of the energy for the stoplight system in the state $\ket{\phi}$, as well as for the new state 

$$ \ket{\eta} = \frac1{\sqrt{3}}\pmatrix{1 \\ 1 \\ 1}. $$

```python
eta = 1/np.sqrt(3)*np.array([1,1,1]);

print("Average energy in phi: ",np.inner(phi,H@phi))
print("Average energy in eta: ",np.inner(eta,H@eta))
```

The Hamiltonian I gave before was very simple because it was **diagonal**, meaning the eigenvalues were just along the line from top left to lower right, and every other entry was 0; however, most operators will not always be so simple. The framework of taking inner products to find probabilities and expectation values will always work, even if the numbers aren't so pretty. Generally, different operators will have different eigenvectors, but the probability of measuring a particular eigenvalue will still depend on the inner product squared of your state with the corresponding eigenvector, and the expectation value will still be the inner product of your state with itself and the operator sandwiched in the middle.

### Practice Problem

Consider an arbitrary new operator for the stoplight system 

$$ \mathcal{A} = \pmatrix{1 & 1 & -3 \\ 1 & 2 & 2 \\ -3 & 2 & -3} $$

 1. Find the eigenvalues and eigenvectors of this operator. 
 2. Find the expectation value of this operator $\langle \mathcal{A} \rangle$ when measuring the state $\ket{\phi}$.
 3. What is the probability of measuring the lowest eigenvalue in an experiment measuring $\mathcal{A}$ on the state $\ket{\phi}$?

```python
A = np.array([ [1,1,-3],[1,2,2],[-3,2,-3] ]);

Avals, Avecs = la.eig(A)
print(Avecs)
print(Avals)

print("Expectation value of A: ",np.inner(phi,A@phi))
print("Probability of measuring ground state: ",np.inner(phi,Avecs[:,0])**2)
```

## Summary

1. Quantum states are vectors which live in a vector space; the number of components in the vector indicates the number of possible measurement outcomes in the system.

2. The probability of measuring a particular outcome is the magnitude squared of the inner product of the quantum state vector $\ket{\Psi}$ and the basis state $\ket{\alpha}$ which corresponds to that measurement outcome:

$$ P(\alpha) = \left|\braket{\Psi}{\alpha}\right|^2 $$

3. Quantum states must be normalized, meaning that the probabilities for all outcomes must add up to 1; in other words, the length of the state vector must equal 1.

4. Quantum operators are matrices which multiply the quantum state vectors. These matrices have orthogonal eigenvectors (eigenstates) which are the basis states for some quantum measurement, and different operators may have different eigenstates. 

5. The eigenvalues which correspond to the eigenvectors of some operator are the allowed, physically observed values when you measure the system. For example, the eigenvalues of the Hamiltonian operator are the allowed values of the energy for the system, and any measurement of the energy will yield one of the eigenvalues with some probability. 

6. The expectation value of some operator in a state is the average value observed in repeated measurements of that value. If I have a 2-state system with energies $0$ and $E$, then the expectation value for a state which has 50/50 probability to be in either energy state will be $E/2$, although you can never measure the energy $E/2$ in a single experiment.
