## What is Spin?

To understand what exactly spin is, there are two pieces of classical physics we need to understand. First, the basic physics of an electromagnet: when a charged object moves in a closed loop, forming a current, it induces a magnetic field through the loop. I can then imagine that an electron traveling in small loops makes its own magnetic field, resembling the dipole field of a bar magnet; that magnet can then be pushed and pulled in the presence of other magnetic fields. There's only one problem with this picture: physicists noticed that electrons seem to have a small dipole field *even while remaining completely still*. Somehow, this tiny magnetic field attached to an electron always exists, created by some internal property we call **spin**. We'll discuss more about these magnetic properties a bit more later.

The second piece of classical (technically, semi-classical) physics we need is a rough understanding of atomic transitions. You've probably seen a picture of an atom with a nucleus in the center and electrons sitting on circular orbits; those orbits are a visualization of the many quantum states the electrons can be in. Early atomic physicists noticed that electrons can move between certain states, but there are other combinations states which electrons can't move between. In order to transition from one state to another, both energy and momentum need to be conserved. Energy conservation is easy; the electron can either emit or absorb photons (light) that carry the energy difference. The complication comes in conserving angular momentum. The orbits which an electron is able to transition between don't all seem to have the same angular momentum, so it seems like angular momentum isn't being conserved. We can resolve this by assuming that the electron has some intrinsic angular momentum due to spin, and then the allowed transitions between atomic states conserve the total angular momentum, adding both the momentum due to the orbit with the built-in spin angular momentum.

In summary: electrons appear to have intrinsic angular momentum and an intrinsic magnetic field, even when stationary in a vacuum. You could try to imagine that these properties are coming from the electric charge on the particle as it rotates about some axis but there are two problems: 

1. An electron at the quantum level is described by a wave function, which is just a cloud of probabilities for measurements. There is no literal ball of charge with a definite size, shape, and position, so how can you describe an axis of rotation for an object which has no shape or orientation? 
2. Even measuring on energy scales where the electron *does* behave like a classical ball of charge, the charge is spread evenly throughout the region where the electron is, and creating a magnetic field requires an imbalance of charge moving around to form a current, something a single electron can't produce on its own.

So now that you're probably thoroughly confused, just trust that these classically unexplainable properties of electrons are due to a new, totally quantum attribute which bears the (unfortunately misleading) name **spin**. As a quantum property, it only comes in discrete values, which we measure in units of Planck's constant $\hbar = h/2\pi \approx 1.05 \cdot 10^{-34} J\cdot s$; notice that this shares the same SI units as angular momentum ($J\cdot s$). For the sake of this project, we'll often just work with units where we set $\hbar=1$ so that we don't have to drag around extra constants in our calculations.

Finally, we describe electrons as spin-1/2 objects (for math reasons too complicated to explain well here), which means that a quantum measurement of spin can be either $\pm\hbar/2$. Other fundamental particles can have any half-integer spin; for example:

 - Photons, the quantum particle of electromagnetic radiation, are spin-1, meaning that a quantum measurement of spin can be $-\hbar, 0, +\hbar$.
 - Protons, neutrons, and their constituent quarks, are also spin-1/2.
 - Nuclei of atoms and other exotic particles made of even numbers of quarks can be any integer spin, and odd numbers of quarks can be any half-integer (3/2, 5/2, 7/2, etc.).

## Spin-1/2 States

For the remainder of this project, we'll only focus on spin-1/2 electrons and the mathematics needed to describe their quantum states. As mentioned previously, there are two possible measurements of spin, meaning I'll need a two-component vector to write down a quantum spin state, with two basis vectors. For historical reasons, the basis vectors are the arrow states from the Quantum Mechanics notebook, labeled by "up" and "down":

$$ \ket{\uparrow} = \pmatrix{1 \\ 0};\ \ket{\downarrow} = \pmatrix{0 \\ 1}. $$

This means for an arbitrary spin state of an electron, I can write it simply as the two-component vector

$$ \ket{\psi} = \pmatrix{\alpha \\ \beta} = \alpha \ket{\uparrow} + \beta \ket{\downarrow} $$

where $\alpha$ and $\beta$ could be any complex numbers.

## Pauli Matrices

As you covered in the previous quantum mechanics notebook, doing a quantum measurement corresponds to multiplying your state vector by an operator (matrix); the eigenvalues of the operator are the values you would measure in an experiment, and the eigenvectors are the states the system could be in after you measure. For spin-1/2 systems, the operators we use to measure are called **Pauli Matrices**, named for Wolfgang Pauli. What we measure in the lab is the component of a spin that points in a given spatial direction (sort of like asking for an axis of rotation), so there are 3 (well, sort of 4) operators, one corresponding to each measurement direction $\hat{x}$, $\hat{y}$, and $\hat{z}$ (and technically, the $2\times 2$ identity matrix is another one, but it's not very interesting to measure). These matrices are defined as follows:

$$ \sigma_x = \pmatrix{0 & 1 \\ 1 & 0};\ \sigma_y = \pmatrix{0 & -i \\ i & 0};\ \sigma_z = \pmatrix{1 & 0 \\ 0 & -1};\ \left[ \sigma_0 = \pmatrix{1 & 0 \\ 0 & 1} \right] $$

*Technically* the spin operators measured in a real experiment are these matrices multiplied by a factor of $\hbar/2$, but we'll ignore that factor so that all of our operators have the same eigenvalues: $\pm 1$. 

### Practice Problem

Using either Python or by hand, determine the eigenvectors of the three matrices $\sigma_x, \sigma_y, \sigma_z$. Keep the matrices and their eigenvectors handy, they'll probably be useful to have at short notice.

```python
PauliX = np.array([[0,1],[1,0]]);
PauliY = np.array([[0, -1j],[1j,0]]);
PauliZ = np.array([[1,0],[0,-1]]);

Xvals, Xvecs = la.eig(PauliX);
Yvals, Yvecs = la.eig(PauliY);
Zvals, Zvecs = la.eig(PauliZ);

print(f"X eigenstates: {Xvecs}")
print(f"Y eigenstates: {Yvecs}")
print(f"Z eigenstates: {Zvecs}")
```

As a matter of convention, you'll usually see the eigenvectors of $\sigma_z$ written as $\ket{0}$ and $\ket{1}$, with the corresponding eigenvalues $+1$ and $-1$ respectively. The eigenvectors of $\sigma_x$ are usually denoted $\ket{+}$ and $\ket{-}$, and for $\sigma_y$, we have $\ket{+i}$ and $\ket{-i}$.

## The Bloch Sphere

<img src="https://www.researchgate.net/profile/Andreas_Ketterer/publication/316828699/figure/fig3/AS:669404337565697@1536609839396/color-online-Bloch-sphere-representation-of-the-Bloch-vector-v-for-qubits-d-2.png" alt="Bloch Sphere" width=500>

A useful way to visualize spin-1/2 states is on the Bloch sphere; in this representation, the spin state is an arrow pointing in a given direction on a sphere. How do we map a general 2-component spin vector of the form $(\alpha, \beta)^\top$ onto a sphere? Since a sphere's surface is a 2-dimensional object, we can describe any position on the sphere by 2 angular coordinates $\theta$ and $\phi$. Since all good quantum states are normalized, we can choose to rewrite the state as:

$$ \ket{\psi} = \pmatrix{\alpha \\ \beta} = \alpha \ket{0} + \beta \ket{1} = \cos(\theta/2)\ket{0} + e^{i\phi}\sin(\theta/2) \ket{1} = \cos(\theta/2) \ket{0} + \left[\cos(\phi) + i \sin(\phi)\right]\sin(\theta/2) \ket{1}. $$

If we associate the $\theta$ and $\phi$ here with coordinates on a sphere, $\theta$ is the angle away from the $\hat{z}$-axis (sometimes called the *colatitude* or *azimuthal angle*), and $\phi$ is the angle away from the $\hat{x}$-axis, measured in the $x-y$ plane (sometimes called the *longitude* or *polar angle*). In regular Cartesian coordinates, the vector $\ket{\psi}$ on the Bloch sphere can be described as the vector $\left(\sin(\theta)\cos(\phi), \sin(\theta)\sin(\phi), \cos(\theta) \right)$, which you may or may not recognize as the usual conversion from Cartesian to spherical coordinates, with the radius of the sphere set to 1.

### Practice Problem

Verify that the state $\ket{\psi}$ above is normalized for all values of $\theta$ and $\phi$, by checking that the product $\bra{\psi}\ket{\psi} = 1$. **Hint:** Remember that $\bra{\psi}$ turns the column vector to a row vector, and the complex conjugate turns all $i$'s into $-i$'s.

$$ \begin{split} 
\ket{\psi} &= \pmatrix{\cos(\theta/2) \\ e^{i\phi} \sin(\theta/2)}; \\
\bra{\psi} \ket{\psi} &= \pmatrix{\cos(\theta/2) & e^{-i\phi}\sin(\theta/2)} \pmatrix{\cos(\theta/2) \\ e^{i\phi} \sin(\theta/2)} \\
&= \cos^2(\theta/2) + \sin^2(\theta/2) \\
&= 1
\end{split} $$

The three coordinate axes for the sphere correspond with the three axes of spin components that we can measure; in this way, the basis states for each spin projection coincide with the basis vectors on the three axes.

 - The states $\ket{0}$ and $\ket{1}$ are on the vectors $\pm \hat{z}$, respectively.
 - The $\sigma_x$ basis states, $\ket{+}$ and $\ket{-}$ are on the vectors $\pm \hat{x}$.
 - Likewise, the $\sigma_y$ basis states, $\ket{+i}$ and $\ket{-i}$ are on the vectors $\pm \hat{y}$.

A few things useful things to know in this picture: 
1.  Orthogonal states (i.e. states which have an inner product of 0) sit on opposite sides of the sphere, $180^\circ$ away from each other; this is opposite our typical intuition that orthogonal vectors are $90^\circ$ away from each other. **Being orthogonal quantum states and being perpendicular on the Bloch sphere are not the same thing!**
2.  The states on the "equator" (including the $\sigma_x$ and $\sigma_y$ eigenstates) are all 50/50 superpositions of the states $\ket{0}$ and $\ket{1}$, so naturally they lie exactly halfway between the "North" and "South" poles.
3.  More generally, when we measure the projection of spin along an axis, the probability of measuring a particular state depends on the angle between the state and the axis in question. For example, a state pointing in the upper hemisphere is more likely to measure $\ket{0}$ than $\ket{1}$, and the expectation value of $\sigma_z$ will be positive, since I will measure the eigenvalue $+1$ in an experiment more often than $-1$.

## Measuring Spin in the Lab: Stern-Gerlach

To observe the effects of spin experimentally, Otto Stern and Walther Gerlach set up an experiment in the early 1920s with the following setup:

1. A source sends a beam of silver atoms outward in a line. Silver has one unpaired valence electron, so the spin of the entire silver atom altogether looks like the spin of a single electron. Nothing special is done to the atoms as they leave the source, so the spin of each atom is oriented completely randomly.
2. The silver atoms travel through a region of magnetic field oriented on the $\hat{z}$-axis; if we imagine each spin as a small, randomly oriented bar magnet, this field will push the atoms upward or downward different amounts, depending how each spin is pointed.
3. The silver atoms collide with a screen where their position is noted.

Now that you've thought about it, I'll spoil the answers. Assuming the silver atoms and their electrons were classical objects, with randomly oriented spins, the field should deflect every atom a slightly different amount, and the screen should show the silver atoms evenly fanned out from top to bottom. **BUT** because spin is quantum, the randomly oriented spins pick one eigenstate to be in when they enter the field, so instead we see two distinct islands of points: one where the spins picked the up eigenstate ($\ket{0}$) and one where they picked the down eigenstate ($\ket{1}$). All the spins are deflected by the same amounts, since the spin value is quantized; either -1 or +1, but nothing between.

You might wonder if the spins simply weren't random to begin with, but they were already labeled somehow as "up" and "down" before reaching the magnetic field. Here's an experiment to check that:

1. Send the silver atoms through the original device, with the magnetic field oriented up-down. As we've seen, this splits the beam in two, one for "up" and one for "down".
2. In front of the "up" spin beam, set up a new magnetic field oriented left-right. This will have a similar effect, forcing the spins to pick either a "plus" or "minus" state, splitting the beam again.
3. In front of the "plus" beam, set up a copy of the magnetic field from step 1, oriented up-down. Recall that every spin that came through step 2 was already measured to have spin up. What result do we see now?

## Summary

 - Spin is a purely quantum, intrinsic property of fundamental particles, which explains the intrinsic magnetic dipole and angular momentum that these particles are observed to have.
 - As a quantized property, spin is only measured in discrete units of $\hbar$. For a spin-1/2 object, I can measure it with spin $\pm \hbar/2$, but never anything else.
 - Similar to an axis of rotation, spin can be decomposed into components along the three Cartesian $\hat{x}, \hat{y}, \hat{z}$ axes. Measuring a component of spin along an axis is mathematically described by the Pauli matrices.
 - Spin-1/2 states are simply 2-component vectors, usually given in the basis of the eigenstates of $\sigma_z$ labeled as $\ket{0}$ and $\ket{1}$.
 - Spin states can be visually described by a Bloch sphere; the eigenstates of all the Pauli matrices lie on their corresponding unit vectors/axes on the sphere.
 - Measuring spin in the lab can be done with a Stern-Gerlach device, which separates spins by their component "up" or "down" along an axis set by the orientation of a magnetic field.
